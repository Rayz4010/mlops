### **Need of numpy is for multi dimensional data**

#### **Now to create a multi dimensional array we just need to make a list of list|**

```python
ma=[[1,2,3],
    [4,5,6],
    [7,8,9]]
```

```python
np.array(ma)
```

```python
np.shape(ma)
```

In numpy we cannot have inconsitent number of elements like

```python
na=[[1,2],
    [4,5,6],
    [7,8,9]]
na
```

but in numpy array we get something like 

```python
np.array(na)
```

arrays can contain any thing like booleans, decimal, strings

```python
s=[['Heallo','Konichiwa'],
    ['Bonjour','Hola']]
np.array(s)
```

can contain mixed

```python
ch=[[1,3],
    [True,4],
    [5,6.0]]
np.array(ch)
```

```python
ch=[[1,'Hello'],
    [True,4],
    [5,6.0]]
np.array(ch)
```

#### **Reshape a array**

```python
x=[[1,2,3],
   [4,5,6]]
p=np.array(x)
np.shape(p)
```

```python
p.reshape(3,2)
```

can be one dimensional too

```python
p.reshape(6,)
```

but this doesnot make any sense

```python
p.reshape(3)
```

### **Now a quick way to build numpy arrays**

```python
np.zeros((3,4))  #3 rows and 4 columns
```

```python
np.ones((2,3))  #3D array with 2 blocks, 3 rows and 4 columns
```

```python
g=np.arange(1,10)
g.reshape(3,3)
```

```python
np.arange(1,10,2)#low high step
```

```python
np.linspace(1,10,num=5)
```
