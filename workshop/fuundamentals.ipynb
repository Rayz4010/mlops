### **Abstraction**

question on abstraction on interview<br>
shows essential details while hiding implementation details<br>
Abstracrt class- A class which has abstract method and which inherits from abc<br>
Abstract method- a method which is decleared but not defined<br>
We cant create obj for abs class<br>
Concrete class: a calss which is inherited from abs class <br>


```python
from abc import ABC,abstractmethod
'''abs class'''
class Abs(ABC):
    @abstractmethod #decoarator
    def abs_method(self):
        pass
'''concrete class'''
class Con(Abs):
    def abs_method(self):
        print('this is implemented method')
        print('this is concrete class method')
        pass
obj=Con()
obj.abs_method()
```

### **Inheritance**

types-single, multiple, hierarchical, multi

#### **Single level Inhertance**

```python
class Parent:
    def func1(self):
        print('this is parent class method')
        pass
class Child(Parent): #inheritance
    def func2(self):
        print('this is child class method')
        pass
obj=Child()
obj.func1()
obj.func2()
```

#### **Multi Level inheritance**

```python
class Grandparent:
    def func3(self):
        print('this is grandparent class method')
        pass
class Parent(Grandparent):
    def func1(self):
        print('this is parent class method')
        pass
class Child(Parent): #inheritance
    def func2(self):
        print('this is child class method')
        pass
obj=Child()
obj.func1()
obj.func2()
obj.func3()
```

#### **Multiple Inheritance**

```python
class Father:
    def func1(self):
        print('this is father class method')
        pass
class Mother:
    def func2(self):
        print('this is mother class method')
        pass
class Sister:
    def func4(self):
        print('this is sister class method')
        pass
class Child(Father,Mother,Sister): #inheritance
    def func3(self):
        print('this is child class method')
        pass
obj=Child()
obj.func1()
obj.func2()
obj.func3()
obj.func4()
```

#### **Hybrid Inheritance**

```python
class Parent:
    def func1(self):
        print('this is parent class method')
        pass
class Child(Parent): #inheritance
    def func2(self):
        print('this is child class method')
        pass
class Child2(Parent):
    def func3(self):
        print('this is child2 class method')
        pass
class Child3(Parent):
    def func4(self):
        print('this is child3 class method')
        pass
c1=Child()
c2=Child2()
c3=Child3()
c1.func1()
c1.func2()
c2.func1()
c2.func3()
c3.func1()
c3.func4()
```
