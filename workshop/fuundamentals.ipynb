## **Object Oriented Programing**

### **Classes and Objects**

```python
class Lol:
    a=10 #class variable
    def load(self):
        print('this is a method')
        pass
    '''self is an argument that call itself'''
obj=Lol() #object
print(obj.a)
#print(Lol.a)
obj.load()
```

### **Constructors**

```python
class Details:
    #defining a constructor
    def __init__(self,name,age):
        self.name=name #isntance variable
        self.age=age
        pass
    def show(self,marks=100):
        print(f'name is {self.name} and age is {self.age} got makrs {marks}')
        pass
obj=Details('ll',23)
obj.show()
obj1=Details('ray',20)
obj1.show(90)
```

### **Scope of variable**

```python
n=1000 #global variable
'''can acess throughout the execution of progrram'''
class Test:
    a=10 #class variable
    def __init__(self):
        self.a=a #instance variable
        pass
    
    def func(self,a):
        b=200 #local variable
        print(n)
        print(self.a)
        print(b)
        pass
obj=Test()
obj.func(100)
```
