## **Heart Disease Prediction**

```python
#import libraries
import pandas as pd
from pandas import read_csv as rc
from sklearn.tree import DecisionTreeClassifier
```

```python
#Load the dataset
df=rc('/Users/rayquaza/Desktop/mlops/workshop/ICE-Workshop/datasets/heart.csv')
df
```

```python
#Split the dataset into x and y
x=df.drop('target',axis=1)
y=df['target']
```

```python
x
```

```python
y
```

```python
#train test split
from sklearn.model_selection import train_test_split as tts
```

```python
x_train,x_test,y_train,y_test=tts(x,y,test_size=0.2,random_state=42) #you can give test_size=0.8
#random_state tends to go for best fit line
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
#call the model and train the model
dt=DecisionTreeClassifier()
dt.fit(x_train,y_train)
```

```python
#make predictions usinfg usrer input
df.columns
```

```python
pred=dt.predict([[53,1,0,140,203,1,0,155,1,3.1,0,0,3]])
pred
```

```python
if pred.item() is 0:
    print('will not die')
else:
    print('will die')
```

```python
#model evaluation
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
```

```python
y_pred=dt.predict(x_test)
acc=accuracy_score(y_test,y_pred)
```

```python
cr=classification_report(y_test,y_pred)
print(cr)
```

```python
conf_matrix=confusion_matrix(y_test,y_pred)
print(conf_matrix)
```
